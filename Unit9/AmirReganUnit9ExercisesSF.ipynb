Answer 1:

For matrix multiplication the rows of the first matrix are aligned with the columns of the second matrix so that both have the same amount of numbers, three. Then the first number of the first row of the first matrix is multiplied with the first number of the first column in the second matrix. All three aligned numbers are multipled and then summed up to get the first number of the final matrix. Then the second column of the second matrix is multiplied with the first row of the first matrix and summed up to make the second value in the final matrix. So on, this creates the whole matrix moving down and multiplying each column with each row to create the six values of the final matrix.
